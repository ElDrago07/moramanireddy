```python
import pandas as pd

df = pd.read_csv(r"E:\downloads\matches.csv")

def find(col_options):
    cols = list(df.columns)
    for opt in col_options:
        for c in cols:
            if c.strip().lower() == opt.strip().lower():
                return c
    for opt in col_options:
        for c in cols:
            if opt.strip().lower() in c.strip().lower():
                return c
    return None

player_col = find(["player_of_match", "player_of", "player_of_m"])
toss_winner_col = find(["toss_winner", "toss_winn", "tosswinn"])
toss_decision_col = find(["toss_decision", "toss_decis"])
winner_col = find(["winner"])
win_runs_col = find(["win_by_runs", "win_by_ru", "win_by_run"])
win_wickets_col = find(["win_by_wickets", "win_by_w"])
venue_col = find(["venue"])
umpire1_col = find(["umpire1"])
umpire2_col = find(["umpire2"])
team1_col = find(["team1"])
team2_col = find(["team2"])
```

```python
print("Q1 Total matches:", len(df))
print("Q1 Columns:", list(df.columns))
print("Q1 First 5 rows:")
print(df.head(5))
print("Q1 Describe:")
print(df.describe(include='all').to_string())
```

```python
if player_col and win_runs_col and win_wickets_col:
    final_ball = df[(df[win_runs_col] == 1) | (df[win_wickets_col] == 1)]
    if not final_ball.empty and player_col in final_ball.columns:
        vc = final_ball[player_col].value_counts()
        if not vc.empty:
            top_player = vc.idxmax()
            top_count = int(vc.max())
            print("Q2 Player with most 'Player of the Match' awards in final-ball games:", top_player, top_count)
        else:
            print("Q2 No data for player_of_match in final-ball games")
```

```python

if venue_col and win_runs_col and win_wickets_col:
    wankhede = df[df[venue_col].str.strip().str.lower() == "wankhede stadium"]
    if not wankhede.empty:
        runs_wins = int((wankhede[win_runs_col] > 0).sum())
        wickets_wins = int((wankhede[win_wickets_col] > 0).sum())
        more_common = "runs" if runs_wins > wickets_wins else ("wickets" if wickets_wins > runs_wins else "equal")
        print("Q3 Wankhede - win by runs:", runs_wins, "win by wickets:", wickets_wins, "more common:", more_common)
    else:
        print("Q3 No matches at Wankhede Stadium in data")
```

```python
if winner_col and win_runs_col:
    big_wins = df[df[win_runs_col] > 50]
    if not big_wins.empty:
        team_counts = big_wins[winner_col].value_counts()
        top_team = team_counts.idxmax()
        top_count = int(team_counts.max())
        print("Q4 Team with most >50 run wins:", top_team, top_count)
    else:
        print("Q4 No wins with margin >50 runs")
```

```python

if toss_winner_col and toss_decision_col and winner_col:
    td = df[toss_decision_col].astype(str).str.lower()
    bat_mask = td.str.contains("bat")
    toss_and_won = df[bat_mask & (df[toss_winner_col] == df[winner_col])]
    print("Q5 Times toss winner batted first and won:", int(len(toss_and_won)))
```

```python
if umpire1_col and umpire2_col and team1_col and team2_col:
    kkr_mask = (df[team1_col].str.strip().str.lower() == "kolkata knight riders") | (df[team2_col].str.strip().str.lower() == "kolkata knight riders")
    kkr_matches = df[kkr_mask]
    u1_count = int(kkr_matches[umpire1_col].notna().sum())
    u2_count = int(kkr_matches[umpire2_col].notna().sum())
    more = umpire1_col if u1_count > u2_count else (umpire2_col if u2_count > u1_count else "equal")
    print("Q6 umpire1 matches with KKR:", u1_count, "umpire2 matches with KKR:", u2_count, "more:", more)
```
